```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = pd.read_csv("sales_data.csv")
data.fillna(value = 0, inplace = True)
data["year"] = pd.to_datetime(data["date"]).dt.year
data["months"] = pd.to_datetime(data["date"]).dt.month
data["day"] = pd.to_datetime(data["date"]).dt.day
_2023_data = data.query("year == 2023")
print(_2023_data["quantity"])
```

```python
y = _2023_data["discount_pct"].reset_index(drop = True)
X = _2023_data[["quantity"]].reset_index(drop = True)
X_train,X_test,y_train,y_test = train_test_split(X, y, test_size = 0.4, random_state = 23)
model  = LinearRegression()
model.fit(X_train,y_train)
print(f" intercept {model.intercept_}")
print(f" coefficient {model.coef_}")
```

```python
y_predict = model.predict(X_test)
predictions = pd.DataFrame({"Actual": y_test, "Predicted": y_predict})
print(predictions.head(5))
```

```python
model.score(X_test,y_test)
```
